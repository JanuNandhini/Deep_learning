Dataset Preparation and Text Preprocessing

```python
print("E.JANU NANDHINI,24BAD041")
!pip -q install nltk gensim

import nltk
import re
import numpy as np
import gensim

from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.probability import FreqDist
from gensim.models import Word2Vec

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

```

```python
category = "news"
sentences = brown.sents(categories=category)

print("Original Sentences:")
for sentence in sentences[:5]:
    print(" ".join(sentence))

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()


processed_sentences = []

for sentence in sentences:
    processed = []

    for word in sentence:
        word = word.lower()


        if word.isalpha() and word not in stop_words:
            word = lemmatizer.lemmatize(word)
            processed.append(word)

    if processed:
        processed_sentences.append(processed)

print("\nPreprocessed Sentences:")
for sentence in processed_sentences[:5]:
    print(" ".join(sentence))

print("\nTotal processed sentences:", len(processed_sentences))
```

Vocabulary Generation and Word Frequency Analysis

```python

all_words = [
    word
    for sentence in processed_sentences
    for word in sentence
]

vocabulary = set(all_words)

frequency_distribution = FreqDist(all_words)

print("Vocabulary Size:", len(vocabulary))
print("Total Word Count:", len(all_words))

print("\nTop 10 Most Frequent Words:")
for word, count in frequency_distribution.most_common(10):
    print(word, ":", count)

print("\nFrequency of selected words:")
for word in ["government", "year", "people"]:
    print(word, ":", frequency_distribution[word])
```

Word Embedding Generation Using Word2Vec

```python

model = Word2Vec(
    sentences=processed_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1,
    workers=4,
    seed=42,
    epochs=20
)

print("Word2Vec training completed.")
print("Embedding dimension:", model.wv.vector_size)
print("Context window:", model.window)
print("Vocabulary size in model:", len(model.wv))
```

```python

selected_word = "government"

if selected_word in model.wv:
    vector = model.wv[selected_word]

    print("Selected Word:", selected_word)
    print("\nEmbedding Vector:")
    print(vector)

    print("\nVector Dimensions:", vector.shape)

    print("\nFive Most Similar Words:")
    for word, score in model.wv.most_similar(
        selected_word, topn=5
    ):
        print(word, ":", round(score, 4))
else:
    print("Selected word is not in the vocabulary.")
```

Word Similarity Analysis and Experiment Management

```python

word_pairs = [
    ("government", "state"),
    ("man", "woman"),
    ("city", "country")
]

similarity_results = []

for word1, word2 in word_pairs:
    if word1 in model.wv and word2 in model.wv:
        score = model.wv.similarity(word1, word2)
        similarity_results.append((word1, word2, score))

        print(f"Word Pair: {word1} - {word2}")
        print("Cosine Similarity:", round(float(score), 4))
        print("-" * 40)
    else:
        print(f"Skipping {word1} - {word2}: word not in vocabulary.")

if similarity_results:
    best_pair = max(similarity_results, key=lambda item: item[2])

    print("\nMost Similar Pair:",
          best_pair[0], "-", best_pair[1])
    print("Highest Similarity Score:",
          round(float(best_pair[2]), 4))

```
